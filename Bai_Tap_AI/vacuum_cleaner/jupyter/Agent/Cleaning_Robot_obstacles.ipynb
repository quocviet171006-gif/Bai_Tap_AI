1. PEAS cho máy hút bụi.
- P: Hàm đánh giá quảng đường ngắn nhất để dọn sạch bụi trong nhà.
- E: Ma trận m x n biểu diễn kích thước của căn phòng, vị trí có rác là 1, sạch là 0, máy hút bụi ở tại vị trí sạch là 2, máy tại ô bẩn là 3.
- A: các trạng thái di chuyển, hành động hút bụi.
- S: cảm biến đồ vật, bụi và camera.

```python
home = [
    [1, 1, 3, 1],
    [0, 1, 1, 1],
    [1, 0, 0, 1]
]
# máy hút bụi ở ô bẩn là 3, ô sạch là 2
# 1 là ô bẩn, 0 là sạch
```

```python
def interpret_input(percept):
    return percept
```

```python
def find_location(state):
    for i in range(len(state)):
        for j in range(len(state[0])):
            if state[i][j] == 2 or state[i][j] == 3:
                return i, j
    return None, None
```

```python
import random

def rule_match(state):
    x, y = find_location(state)
    
    if state[x][y] == 3:
        return "Suck"
    
    move_options = []
    if x > 0: move_options.append("Up")
    if x < len(state) - 1: move_options.append("Down")
    if y > 0: move_options.append("Left")
    if y < len(state[0]) - 1: move_options.append("Right")
    
    return random.choice(move_options)
```

```python
def do_action(state, action):
    x, y = find_location(state)
    rows = len(state)
    cols = len(state[0])
    
    if action == "Suck":
        if state[x][y] == 3:
            state[x][y] = 2
        return state

    nx, ny = x, y
    if action == "Up": nx -= 1
    elif action == "Down": nx += 1
    elif action == "Left": ny -= 1
    elif action == "Right": ny += 1

    if not (0 <= nx < rows and 0 <= ny < cols):
        return state 
    
    if state[nx][ny] == -1:
        return state

    state[x][y] = 1 if state[x][y] == 3 else 0
    
    if state[nx][ny] == 1:
        state[nx][ny] = 3
    else:
        state[nx][ny] = 2
        
    return state
```

```python
def simple_reflex_agent(percept):
    state = interpret_input(percept)
    action = rule_match(state)
    print(f"Hành động: {action}")
    return do_action(state, action)

current_state = home
print("Trạng thái ban đầu:")
for row in current_state: print(row)

for i in range(1, 11):
    print(f"\n--- Bước {i} ---")
    current_state = simple_reflex_agent(current_state)
    for row in current_state:
        print(row)
```
