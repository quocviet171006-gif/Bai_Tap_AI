```python
matrix = [
    [1, 4, 5],
    [2, 0, 6],
    [3, 8, 7]
]

for row in matrix:
    print(row)
```

```python
# tìm vị trí ô trống
def find_blank(state):
    for i in range(3):
        for j in range(3):
            if state[i][j] == 0:
                return i, j

```

```python
def rule_match(state):
    x, y = find_blank(state)
    
    move_options = []
    if x > 0: move_options.append("Up")
    if x < len(state) - 1: move_options.append("Down")
    if y > 0: move_options.append("Left")
    if y < len(state[0]) - 1: move_options.append("Right")
    
    return move_options
```

```python
def create_copy(matrix):
    new_matrix = []

    for i in range(len(matrix)):
        row = []
        for j in range(len(matrix[i])):
            row.append(matrix[i][j])
        new_matrix.append(row)
    return new_matrix
```

```python
def is_equal(matrix_1, matrix_2):
    for i in range(len(matrix_1)):
        for j in range(len(matrix_1[i])):
            if matrix_1[i][j] != matrix_2[i][j]:
                return False
    return True
```

```python
def do_action(state, action):
    # Tạo một bản sao mới hoàn toàn để không ảnh hưởng đến state gốc
    new_state = create_copy(state)
    x, y = find_blank(new_state)

    if action == "Right":
        new_state[x][y], new_state[x][y + 1] = new_state[x][y + 1], new_state[x][y]
    elif action == "Left":
        new_state[x][y], new_state[x][y - 1] = new_state[x][y - 1], new_state[x][y]
    elif action == "Up":
        new_state[x][y], new_state[x - 1][y] = new_state[x - 1][y], new_state[x][y]
    elif action == "Down":
        new_state[x][y], new_state[x + 1][y] = new_state[x + 1][y], new_state[x][y]

    return new_state
```

```python
import random

def model_based_reflex_agent(percept):
    global history

    # 1. Nhận diện trạng thái hiện tại (Model update)
    current_state = create_copy(percept)
    
    # 2. Đưa trạng thái hiện tại vào bộ nhớ nếu chưa có
    is_new = True
    for s in history:
        if is_equal(current_state, s):
            is_new = False
            break
    if is_new:
        history.append(create_copy(current_state))

    # 3. Lấy luật từ môi trường
    all_possible_moves = rule_match(current_state)
    fresh_moves = []
    
    for action in all_possible_moves:
        # Giả lập bước đi tiếp theo
        test_state = create_copy(current_state)
        next_state = do_action(test_state, action)

        # Kiểm tra xem bước đi này đã từng thử chưa
        visited = False
        for old in history:
            if is_equal(next_state, old):
                visited = True
                break
        if not visited:
            fresh_moves.append(action)

    # 4. Chọn hành động
    if fresh_moves:
        final_action = random.choice(fresh_moves)
    else:
        # Nếu không có nước mới, chọn ngẫu nhiên để thoát kẹt
        final_action = random.choice(all_possible_moves)
    
    # Thực hiện hành động trên bản sao của ma trận hiện tại
    return do_action(create_copy(current_state), final_action)
```

```python
goal_state = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 0]
]

history = [] 
current_matrix = matrix
step = 0
max_steps = 50

print("Trạng thái bắt đầu:")
for r in current_matrix: print(r)

while not is_equal(current_matrix, goal_state) and step < max_steps:
    step += 1
    
    current_matrix = model_based_reflex_agent(current_matrix)
    
    print(f"\nBước {step}:")
    for row in current_matrix:
        print(row)
```
