```python
matrix = [
    [1, 2, 3],
    [4, 5, 6],
    [8, 0, 7]
]

goal_state = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 0]
]
```

```python
class Node:
    def __init__(self, state, parent=None, action=None, g=0):
        self.state = state
        self.parent = parent    # Lưu vết cha để truy xuất đường đi
        self.action = action    # Hành động dẫn đến trạng thái này
        self.g = g              # Số bước đã đi đến hiện tại

    def __eq__(self, other):
        return self.state == other.state
```

```python
# tìm vị trí ô trống
def find_blank(state):
    for i in range(3):
        for j in range(3):
            if state[i][j] == 0:
                return i, j

```

```python
def rule_match(state):
    x, y = find_blank(state)
    
    move_options = []
    if x > 0: move_options.append("Up")
    if x < 2: move_options.append("Down")
    if y > 0: move_options.append("Left")
    if y < 2: move_options.append("Right")
    
    return move_options
```

```python
def do_action(state, action):
    new_state = copy.deepcopy(state)
    
    x, y = find_blank(new_state)

    if action == "Up":
        new_state[x][y], new_state[x-1][y] = new_state[x-1][y], new_state[x][y]
    elif action == "Down":
        new_state[x][y], new_state[x+1][y] = new_state[x+1][y], new_state[x][y]
    elif action == "Left":
        new_state[x][y], new_state[x][y-1] = new_state[x][y-1], new_state[x][y]
    elif action == "Right":
        new_state[x][y], new_state[x][y+1] = new_state[x][y+1], new_state[x][y]

    return new_state
```

```python
import copy
from collections import deque

def solve_puzzle_bfs(start_state, target_state):

    start_node = Node(start_state, g=0)
    
    # Kiểm tra nếu trạng thái đầu là đích luôn
    if start_node.state == target_state:
        return start_node

    frontier = deque([start_node])
    reached = [start_node.state]

    while frontier:
        # Lấy node đầu tiên trong queue ra
        node = frontier.popleft()

        # Duyệt qua các hành động có thể
        for action in rule_match(node.state):
            child_state = do_action(node.state, action)
            
            # Tạo node con với g = g_cha + 1
            child_node = Node(child_state, node, action, g=node.g + 1)

            if child_state == target_state:
                return child_node

            # Nếu state chưa từng khám phá thì thêm vào hàng chờ
            if child_state not in reached:
                reached.append(child_state)
                frontier.append(child_node)
                
    return None
```

```python
def get_path(node):
    path = []
    curr = node
    while curr.parent is not None:
        path.append((curr.action, curr.state))
        curr = curr.parent
    return path[::-1]

print("--- ĐANG GIẢI PUZZLE ---")
result_node = solve_puzzle_bfs(matrix, goal_state)

if result_node:
    path_data = get_path(result_node)
    print(f"Thành công! Tổng số bước đi : {result_node.g}")
    print("-" * 30)
    
    # Hiển thị trạng thái bắt đầu
    print("Trạng thái bắt đầu:")
    for row in matrix: print(row)
    
    # Hiển thị từng bước đi
    for i, (act, st) in enumerate(path_data):
        print(f"\nBước {i+1}: {act}")
        for row in st:
            print(row)
else:
    print("Không tìm thấy lời giải sau khi duyệt hết các khả năng.")
```
