```python
matrix = [
    [1, 4, 5],
    [2, 0, 6],
    [3, 8, 7]
]

for row in matrix:
    print(row)
```

```python
def interpret_input(percept):
    return percept
```

```python
# tìm vị trí ô trống
def find_blank(state):
    for i in range(3):
        for j in range(3):
            if state[i][j] == 0:
                return i, j

```

```python
import random

def rule_match(state):
    x, y = find_blank(state)
    
    move_options = []
    if x > 0: move_options.append("Up")
    if x < len(state) - 1: move_options.append("Down")
    if y > 0: move_options.append("Left")
    if y < len(state[0]) - 1: move_options.append("Right")
    
    return random.choice(move_options)


```

```python
def do_action(state, action):

    x, y = find_blank(state)

    if action == "Right":
        state[x][y], state[x][y + 1] = state[x][y + 1], state[x][y]

    elif action == "Left":
        state[x][y], state[x][y - 1] = state[x][y - 1], state[x][y]

    elif action == "Up":
        state[x][y], state[x - 1][y] = state[x - 1][y], state[x][y]

    elif action == "Down":
        state[x][y], state[x + 1][y] = state[x + 1][y], state[x][y]

    return state
```

```python
def simple_reflex_agent(percept):

    state = interpret_input(percept)
    rule = rule_match(state)
    action = rule

    return do_action(state, action)
```

```python
# 1. Định nghĩa trạng thái đích để so sánh
goal_state = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 0]
]

# 2. Khởi tạo ma trận ban đầu
current_matrix = matrix

print("Trạng thái bắt đầu:")
for r in current_matrix: print(r)

step = 0
max_steps = 50 

while current_matrix != goal_state and step < max_steps:
    step += 1
    
    current_matrix = simple_reflex_agent(current_matrix)
    
    print(f"\nBước {step}:")
    for row in current_matrix:
        print(row)

if current_matrix == goal_state:
    print("\n--- Chúc mừng! Agent đã đưa về trạng thái đích ---")
else:
    print("\n--- Dừng lại do đạt giới hạn bước đi ---")
```
